### Questions

1. What happens to Alice's original qubit?
2. Which qubit should now contain the teleported state?
3. Why are two classical bits required?

<details>
<summary>Show solution</summary>

1. Alice's original qubit is measured during the teleportation protocol.<br>
   The measurement destroys the original quantum state on Alice's side. After the measurement, Alice no longer possesses the state $|\psi\rangle$.<br>
   This is why quantum teleportation does not create a copy of the state.

2. Bob's qubit (q2 in this lab) should contain the teleported state.<br>
   After Alice performs her measurements and Bob applies the appropriate correction (`I`, `X`, `Z`, or `XZ`), q2 becomes:
   $|\psi\rangle$ which is exactly the state that was initially stored in q0.

3. Alice's measurement produces four possible outcomes (00, 01, 10, 11).<br>
   These four possibilities correspond to four different situations for Bob's qubit.<br>
   Since there are four possible outcomes, Alice must send enough information to distinguish between them.<br>
   Two classical bits can represent exactly four values.<br>
   Therefore, two classical bits are required so that Bob knows which correction must be applied to recover the original quantum state.
