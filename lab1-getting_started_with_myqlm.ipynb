## Helper functions

```python
from qutip import Qobj, Bloch

def plot_from_result(result):
    """Plot the Bloch sphere representation of the quantum state from the result.

    Args:
        result: The Result object containing the raw data.
    """
    amplitudes = []
    for sample in result.raw_data:
        amp = complex(
        float(sample._amplitude.re),
        float(sample._amplitude.im)
        )
        amplitudes.append(amp)
    statevector = np.array(amplitudes)
    print("State vector:")
    print(statevector)
    psi = Qobj(statevector.reshape((-1, 1)))
    bloch = Bloch()
    bloch.add_states(psi)
    bloch.show()
```
